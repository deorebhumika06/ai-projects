```
import pandas as pd
data={
    "stud_id":[1,2,3,4,5],
    "name":["bhumika","khushi","raj","harsh","arjun"],
    "java":[91,92,93,94,95],
    "python":[92,93,94,95,96]
}
df=pd.DataFrame(data)
print(df)
```

```
from matplotlib import pyplot
x=[10,20,30,40,50]
y=[100,200,300,400,500]
pyplot.scatter(x,y)
pyplot.show()
```

```
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
sns.histplot(df["java"] , kde=True)
plt.title("Java Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.tight_layout()

plt.subplot(1,2,2)
sns.histplot(df["python"] , kde=True)
plt.title("Python Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
