```python
import re

def extract_Show(Show):
    Show_num = re.findall('[0-9]+', str(Show))  # + use kar
    if len(Show_num) > 0:
        return int(Show_num[0])
    else:
        return None

df['Shows'] = df['Shows'].apply(extract_Show)
```

```python
df.head()
```

```python
import re
def extract_Ref(ref):
  ref_no=re.findall('[0-9]+',str(ref))
  if len(ref_no)>0:
    return int(ref_no[0])
  else:
    return None
df['Ref.']=df['Ref.'].apply(extract_Ref)
```

```python
df.head()
```

```python
df['Ref.'].unique()
```

```python
import re
def extract_Year(year):
  year_num=re.findall('[0-9]+',str(year))
  if len(year_num) >0:
    return int(year_num[0])
  else:
    return None
df['Year(s)']=df['Year(s)'].apply(extract_Year)
```

```python
df
```

```python
df['Year(s)'].unique()
```

```python
year = df['Year(s)'].median()
df['Year(s)'] = df['Year(s)'].fillna(year)
```

```python
df['Tour title'].unique()
```

```python
df['Tour title'].isnull().sum()
```

```python
name = df['Tour title'].mode()[0]
df['Tour title'] = df['Tour title'].fillna(name)
```

```python
df['Tour title'].value_counts()
```

```python
import re

def clean_tour(name):
    name = re.sub(r'\[.*?\]', '', str(name))
    name = re.sub(r'[†‡]', '', name)
    return name.strip()

df['Tour title'] = df['Tour title'].apply(clean_tour)
```
