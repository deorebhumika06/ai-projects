```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns




df_processed = df.copy()

X = df_processed.drop(columns=['Loan_ID', 'Loan_Status']) # Drop Loan_ID as it's an identifier
y = df_processed['Loan_Status']


X['Dependents'] = X['Dependents'].replace('3+', '3').astype(float)


numerical_cols = X.select_dtypes(include=['number', 'float']).columns
categorical_cols = X.select_dtypes(include=['object']).columns


for col in numerical_cols:
    if X[col].isnull().any():
        imputer = SimpleImputer(strategy='median')
        X[col] = imputer.fit_transform(X[[col]])[:, 0]

# Impute missing categorical values using the most frequent value (mode)
for col in categorical_cols:
    if X[col].isnull().any():
        imputer = SimpleImputer(strategy='most_frequent')
        X[col] = imputer.fit_transform(X[[col]])[:, 0]

# One-hot encode categorical features
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# --- Preprocessing for Target (y) ---

# Encode target variable 'Loan_Status' (Y/N) to numerical (0/1)
le = LabelEncoder()
y_encoded = le.fit_transform(y) # 'N' -> 0, 'Y' -> 1 (or vice-versa depending on lexical order)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# --- Decision Tree Model ---
print("\n--- Decision Tree Classifier ---")
dtc_model = DecisionTreeClassifier(random_state=42)
dtc_model.fit(X_train, y_train)

y_pred_dtc = dtc_model.predict(X_test)

print(f"Accuracy (Decision Tree): {accuracy_score(y_test, y_pred_dtc):.2f}")
print("\nClassification Report (Decision Tree):")
print(classification_report(y_test, y_pred_dtc))

print("\nConfusion Matrix (Decision Tree):")
cm_dtc = confusion_matrix(y_test, y_pred_dtc)
plt.figure(figsize=(6, 4))
sns.heatmap(cm_dtc, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (Decision Tree)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# --- Random Forest Model ---
print("\n--- Random Forest Classifier ---")
rfc_model = RandomForestClassifier(random_state=42, n_estimators=100) # Added n_estimators
rfc_model.fit(X_train, y_train)

y_pred_rfc = rfc_model.predict(X_test)

print(f"Accuracy (Random Forest): {accuracy_score(y_test, y_pred_rfc):.2f}")
print("\nClassification Report (Random Forest):")
print(classification_report(y_test, y_pred_rfc))

print("\nConfusion Matrix (Random Forest):")
cm_rfc = confusion_matrix(y_test, y_pred_rfc)
plt.figure(figsize=(6, 4))
sns.heatmap(cm_rfc, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (Random Forest)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# --- Comparison ---
print("\n--- Model Comparison ---")
print(f"Decision Tree Accuracy: {accuracy_score(y_test, y_pred_dtc):.2f}")
print(f"Random Forest Accuracy: {accuracy_score(y_test, y_pred_rfc):.2f}")
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost as xgb

print("--- Step 1: Load the Dataset ---")
# Load the loan prediction dataset
df_loan = pd.read_csv("/content/train_u6lujuX_CVtuZ9i (1).csv")
print("Dataset loaded successfully. Displaying first 5 rows:")
display(df_loan.head())

print("\n--- Step 2: Prepare Features (X) and Target (y) ---")
# Create a copy for processing to avoid modifying the original dataframe
df_processed = df_loan.copy()

# Define features (X) by dropping 'Loan_ID' (identifier) and 'Loan_Status' (target)
X = df_processed.drop(columns=['Loan_ID', 'Loan_Status'])
# Define target (y)
y = df_processed['Loan_Status']
print(f"Shape of features (X): {X.shape}")
print(f"Shape of target (y): {y.shape}")

print("\n--- Step 3: Preprocessing Features ---")
# Handle 'Dependents' column: replace '3+' with '3' and convert to float
# This step ensures the column is numeric for imputation if needed
X['Dependents'] = X['Dependents'].replace('3+', '3').astype(float)

# Identify numerical and categorical columns for separate imputation and encoding
numerical_cols = X.select_dtypes(include=['number', 'float']).columns
categorical_cols = X.select_dtypes(include=['object']).columns

print("  Imputing missing numerical values (using median)...")
# Impute missing numerical values using the median strategy
for col in numerical_cols:
    if X[col].isnull().any():
        imputer_num = SimpleImputer(strategy='median')
        X[col] = imputer_num.fit_transform(X[[col]])[:, 0]

print("  Imputing missing categorical values (using mode)...")
# Impute missing categorical values using the most frequent value (mode) strategy
for col in categorical_cols:
    if X[col].isnull().any():
        imputer_cat = SimpleImputer(strategy='most_frequent')
        X[col] = imputer_cat.fit_transform(X[[col]])[:, 0]

print("  One-hot encoding categorical features...")
# One-hot encode categorical features to convert them into a numerical format
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
print(f"Shape of features (X) after preprocessing: {X.shape}")

print("\n--- Step 4: Encode Target Variable ---")
# Encode target variable 'Loan_Status' (Y/N) to numerical (0/1)
# 'Y' often maps to 1 and 'N' to 0, or vice-versa depending on lexical order.
le = LabelEncoder()
y_encoded = le.fit_transform(y)
print(f"Original target classes: {le.classes_}")
print(f"Encoded target values (first 5): {y_encoded[:5]}")

print("\n--- Step 5: Split Data into Training and Testing Sets ---")
# Split the preprocessed data into 80% training and 20% testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}, y_test shape: {y_test.shape}")

# --- Model Training and Evaluation ---

# Prepare lists to store model names and their accuracies for comparison
model_names = []
accuracies = []

print("\n--- Step 6: Train and Evaluate Decision Tree Classifier ---")
# Initialize and train the Decision Tree Classifier
dtc_model = DecisionTreeClassifier(random_state=42)
dtc_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_dtc = dtc_model.predict(X_test)

# Evaluate the Decision Tree model
dtc_accuracy = accuracy_score(y_test, y_pred_dtc)
print(f"Accuracy (Decision Tree): {dtc_accuracy:.2f}")
print("\nClassification Report (Decision Tree):")
print(classification_report(y_test, y_pred_dtc))

# Display Confusion Matrix for Decision Tree
print("Confusion Matrix (Decision Tree):")
cm_dtc = confusion_matrix(y_test, y_pred_dtc)
plt.figure(figsize=(6, 4))
sns.heatmap(cm_dtc, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (Decision Tree)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
model_names.append('Decision Tree')
accuracies.append(dtc_accuracy)

print("\n--- Step 7: Train and Evaluate Random Forest Classifier ---")
# Initialize and train the Random Forest Classifier
rfc_model = RandomForestClassifier(random_state=42, n_estimators=100) # Added n_estimators
rfc_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_rfc = rfc_model.predict(X_test)

# Evaluate the Random Forest model
rfc_accuracy = accuracy_score(y_test, y_pred_rfc)
print(f"Accuracy (Random Forest): {rfc_accuracy:.2f}")
print("\nClassification Report (Random Forest):")
print(classification_report(y_test, y_pred_rfc))

# Display Confusion Matrix for Random Forest
print("Confusion Matrix (Random Forest):")
cm_rfc = confusion_matrix(y_test, y_pred_rfc)
plt.figure(figsize=(6, 4))
sns.heatmap(cm_rfc, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (Random Forest)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
model_names.append('Random Forest')
accuracies.append(rfc_accuracy)

print("\n--- Step 8: Train and Evaluate XGBoost Classifier ---")
# Initialize and train the XGBoost Classifier
# Using use_label_encoder=False and eval_metric='logloss' to suppress warnings
model_gbc = xgb.XGBClassifier(objective='binary:logistic', n_estimators=100, random_state=42,
                             use_label_encoder=False, eval_metric='logloss')
model_gbc.fit(X_train, y_train)

# Make predictions on the test set
y_pred_gbc = model_gbc.predict(X_test)

# Evaluate the XGBoost model
gbc_accuracy = accuracy_score(y_test, y_pred_gbc)
print(f"Accuracy (XGBoost): {gbc_accuracy:.2f}")
print("\nClassification Report (XGBoost):")
print(classification_report(y_test, y_pred_gbc))

# Display Confusion Matrix for XGBoost
print("Confusion Matrix (XGBoost):")
cm_gbc = confusion_matrix(y_test, y_pred_gbc)
plt.figure(figsize=(6, 4))
sns.heatmap(cm_gbc, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (XGBoost)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
model_names.append('XGBoost')
accuracies.append(gbc_accuracy)

print("\n--- Step 9: Compare Model Accuracies ---")
comparison_df = pd.DataFrame({'Model': model_names, 'Accuracy': accuracies})
display(comparison_df.sort_values(by='Accuracy', ascending=False))

# Optional: Visualize model comparison
plt.figure(figsize=(10, 6))
sns.barplot(x='Model', y='Accuracy', data=comparison_df, palette='viridis')
plt.title('Model Accuracy Comparison')
plt.ylim(0.5, 0.85) # Adjust y-limit based on expected accuracy range
plt.ylabel('Accuracy Score')
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

print("--- Step 1: Load and Inspect Data ---")

display(df.head())
print("\nDataFrame Information:")
df.info()
```

### Step 2: Data Preprocessing

*   **Target Variable**: Discretize `mpg` into categories (`low`, `medium`, `high`) to turn it into a classification problem. This is a common approach when the user asks for classification models (Decision Tree Classifier, Random Forest Classifier) on a dataset with a continuous target.
*   **Handle Missing Values**: Impute `horsepower` using the median.
*   **Categorical Features**: One-hot encode the `origin` column.
*   **Drop unnecessary columns**: Drop the `name` column as it's a unique identifier and not directly useful for this basic model.

```python
# Make a copy to avoid modifying the original DataFrame
df_processed = df.copy()

# Discretize 'mpg' into categories for classification
df_processed['mpg_category'] = pd.qcut(df_processed['mpg'], q=3, labels=['low', 'medium', 'high'])

# Define features (X) and target (y)
X = df_processed.drop(columns=['name', 'mpg', 'mpg_category'])
y = df_processed['mpg_category']

# Impute missing 'horsepower' values with the median
imputer_num = SimpleImputer(strategy='median')
X['horsepower'] = imputer_num.fit_transform(X[['horsepower']])[:, 0]

# One-hot encode 'origin' (categorical feature)
X = pd.get_dummies(X, columns=['origin'], drop_first=True)

# Encode target variable 'mpg_category'
le = LabelEncoder()
y_encoded = le.fit_transform(y)

print(f"Original 'mpg' categories: {le.classes_}")
print(f"Shape of features (X) after preprocessing: {X.shape}")
print(f"Shape of target (y) after encoding: {y_encoded.shape}")
display(X.head())
```

### Step 3: Split Data into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")
```

### Step 4: Decision Tree Classifier Model

```python
dtc_model = DecisionTreeClassifier(random_state=42)
dtc_model.fit(X_train, y_train)

y_pred_dtc = dtc_model.predict(X_test)

print(f"Accuracy (Decision Tree): {accuracy_score(y_test, y_pred_dtc):.2f}")
print("\nClassification Report (Decision Tree):")
print(classification_report(y_test, y_pred_dtc, target_names=le.classes_))

plt.figure(figsize=(6, 4))
sns.heatmap(confusion_matrix(y_test, y_pred_dtc), annot=True, fmt='d', cmap='Blues',
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (Decision Tree)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

### Step 5: Random Forest Classifier Model

```python
rfc_model = RandomForestClassifier(random_state=42, n_estimators=100)
rfc_model.fit(X_train, y_train)

y_pred_rfc = rfc_model.predict(X_test)

print(f"Accuracy (Random Forest): {accuracy_score(y_test, y_pred_rfc):.2f}")
print("\nClassification Report (Random Forest):")
print(classification_report(y_test, y_pred_rfc, target_names=le.classes_))

plt.figure(figsize=(6, 4))
sns.heatmap(confusion_matrix(y_test, y_pred_rfc), annot=True, fmt='d', cmap='Blues',
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix (Random Forest)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
