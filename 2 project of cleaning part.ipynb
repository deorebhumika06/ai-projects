```python
import re
```

```python
def export_duration(duration):
  duration_count=re.findall('[0-9]+',str(duration))
  if len(duration_count) > 0:
    return int(duration_count[0])
  else:
    return None
df['duration']=df['duration'].apply(export_duration)
```
